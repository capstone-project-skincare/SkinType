```python
import os
import cv2
import random
from tqdm import tqdm
import albumentations as A
```

```python
base_path = "SkinType_Dataset/train"
classes = ["Dry", "Normal", "Oily"]
target_count = 450 
```

```python
# Augmentation pipeline
augment = A.Compose([
    A.HorizontalFlip(p=0.5),
    A.VerticalFlip(p=0.3),
    A.RandomBrightnessContrast(p=0.5),
    A.Rotate(limit=20, p=0.5),
    A.RandomGamma(p=0.3),
    A.GaussNoise(p=0.3),
    A.Blur(p=0.2),
])
```

```python
for cls in classes:
    folder = os.path.join(base_path, cls)
    images = [f for f in os.listdir(folder) if f.lower().endswith(('.jpg','.png','.jpeg'))]
    current_count = len(images)
    print(f"{cls}: {current_count} images")

    if current_count >= target_count:
        print(f"✅ {cls} already has enough images.")
        continue

    needed = target_count - current_count
    print(f"➡️ Augmenting {needed} images for {cls}...")

    for i in tqdm(range(needed)):
        img_name = random.choice(images)
        img_path = os.path.join(folder, img_name)
        image = cv2.imread(img_path)

        # apply augmentation
        augmented = augment(image=image)['image']

        # create new filename
        new_name = f"{cls}_aug_{i}.jpg"
        cv2.imwrite(os.path.join(folder, new_name), augmented)

print("✅ Augmentation complete!")
```
